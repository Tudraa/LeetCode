```python
class MinStack(object):

    def __init__(self):
        self.stack = []
        self.minStack = []

    def push(self, value):
        """
        :type value: int
        :rtype: None
        """
        self.stack.append(value) 
        minVal = min(value,self.minStack[-1] if self.minStack else value)   
        self.minStack.append(minVal)

    def pop(self):
        """
        :rtype: None
        """
        self.stack.pop()
        self.minStack.pop()

    def top(self):
        """
        :rtype: int
        """
        return self.stack[-1]

    def getMin(self):
        """
        :rtype: int
        """

        return self.minStack[-1]
        


# Your MinStack object will be instantiated and called as such:

obj = MinStack()
minStack = MinStack()

minStack.push(-2)
minStack.push(0)
minStack.push(-3)
minStack.getMin() 
minStack.pop()
minStack.top()
minStack.getMin() 

minStack
```
