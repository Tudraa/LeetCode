```python
class Solution(object):
    def evalRPN(self, tokens):
        """
        :type tokens: List[str]
        :rtype: int
        """
        operand = ["+", "-", "*", "/"]
        stack = []
        for i in tokens:
            if i not in operand:
                stack.append(int(i))
            else : 
                res = 0
                if i == "+" : 
                    res = stack[-2] + stack[-1]            
                elif i == "-" :
                    res = stack[-2] - stack[-1]           
                elif i == "*" :
                    res = stack[-2] * stack[-1]          
                elif i == "/" :
                    res = int(stack[-2] / stack[-1])          
                else :
                    print ("hallo")
                stack.pop()
                stack.pop()
                stack.append(res)

        return stack[-1]
tokens = ["10","6","9","3","+","-11","*","/","*","17","+","5","+"]
output = Solution()
output.evalRPN(tokens)
```
