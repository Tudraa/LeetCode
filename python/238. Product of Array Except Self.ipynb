### With Division

```python
num = [1,2,3,4]
res = []
sum = 1

for i in num:
    sum = i*sum 
for i in num:
    res.append(sum//i)

print(res)
```

### With Prefix and Sufix

```python
num = [1,2,3,4]
n = len(num)
prefix = [1]*n
suffix = [1]*n
sum = 1

# 1. Hitung Prefix
for i in range(1, n):
    prefix[i] = num[i - 1] * prefix[i - 1]

# 2. Hitung Suffix
for i in range(n - 2, -1, -1):
    suffix[i] = num[i + 1] * suffix[i + 1]

# 3. Gabungkan hasil
for i in range(n):
    res[i] = prefix[i] * suffix[i]

print(res)  # Output: [24, 12, 8, 6]
```
