```python
class Solution:
    def carFleet(self, target: int, position: list[int], speed: list[int]) -> int:
        position_speed = [] #pair [position , speed]
        stack = []

        for i, p in enumerate(position) :
            position_speed.append([p, speed[i]])

        position_speed.sort(reverse=True) # sorted based on position

        max_num = 0
        if not position : return 0

        count = 1
        for pair in (position_speed):
            time_arrived = ((target - pair[0])/pair[1])
            if stack and time_arrived > stack[0]:
                count = count + 1
                stack=[]

            stack.append(time_arrived)
        return count



target = 10
position = [6,8]
speed = [3,2]

ans = Solution()
ans.carFleet(target, position, speed)
```
