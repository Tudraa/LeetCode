```python
import re
s = "A man, a plan, a canal: Panama"
```

```python
class Solution:
    def isPalindrome(self, s: str) -> bool: #O(n/2)
        s = s.lower()
        s = re.sub(r"[^A-Za-z0-9]","",s)
        for c in range(int(len(s)/2)) : 
            # print(s[-1-c])
            if s[c] != s[-1-c] :
                return False
        return True
```

```python
solusi = Solution()
solusi.isPalindrome(s)
```
