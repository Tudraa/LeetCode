```python
class Solution:
    def twoSum(self, numbers: list[int], target: int) -> list[int]:
        l = 0
        r = len(numbers) - 1

        while l < r : 
            sum = numbers[l] + numbers[r]
            if  sum > target :
                r -= 1
            elif sum < target:
                l += 1
            else : 
                break
        
        print(l, r)

numbers = [-1, 0]
target = 6

res = Solution()
res.twoSum(numbers, target)
```
